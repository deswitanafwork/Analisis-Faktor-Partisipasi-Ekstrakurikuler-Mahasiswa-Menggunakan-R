```r
# Menghitung ukuran sampel (n)
n <- nrow(data)

# Menghitung jumlah siswa yang berpartisipasi dalam kegiatan ekstrakurikuler
success_count <- sum(data$ExtraCurricularActivities == "Yes")

# Menghitung proporsi sampel
p_hat <- success_count / n

# Menghitung Interval Kepercayaan 95%
z_value <- 1.96 # untuk 95% confidence level
se <- sqrt(p_hat * (1 - p_hat) / n)
lower_ci <- p_hat - z_value * se
upper_ci <- p_hat + z_value * se

# Uji Hipotesis
pi_0 <- 0.5
z_stat <- (p_hat - pi_0) / sqrt(pi_0 * (1 - pi_0) / n)

# Menghitung Chi-Square
chisquare_stat <- z_stat^2

# Menghitung p-value
p_value <- 1 - pnorm(z_stat)

# Kesimpulan
if (p_value < 0.05) {
  conclusion <- "Tolak H0, proporsi partisipasi lebih dari 50%"
} else {
  conclusion <- "Gagal menolak H0, tidak ada bukti bahwa proporsi lebih dari 50%"
}

# Tampilkan hasil
list(
  response = "ExtracurricularActivities (0 = Not Participating, 1 = Participating)",
  sample_size = n,
  success_category = "Participating",
  sample_proportion = p_hat,
  confidence_interval = c(lower_ci, upper_ci),
  z_statistic = z_stat,
  chisquare_statistic = chisquare_stat,
  p_value = p_value,
  conclusion = conclusion
)
```

```r
# Mendefinisikan keberhasilan (Success) sebagai partisipasi dalam kegiatan ekstrakurikuler
# Di sini kita menganggap partisipasi yang berhasil adalah jika siswa berpartisipasi
data <- data %>%
  mutate(Success = ifelse(ExtraCurricularActivities == "Yes", 1, 0))

# Fungsi untuk menghitung proporsi keberhasilan berdasarkan variabel kategorikal
calculate_proportion <- function(data, predictor) {
  data %>%
    group_by(!!sym(predictor)) %>%  # Mengelompokkan data berdasarkan variabel prediktor
    summarise(Proportion_Success = mean(Success)) %>%  # Menghitung rata-rata keberhasilan (proporsi)
    arrange(desc(Proportion_Success))  # Mengurutkan hasil dari proporsi tertinggi
}

# Menghitung proporsi keberhasilan untuk setiap prediktor kategorikal
proportion_gender <- calculate_proportion(data, "Gender")            # Proporsi berdasarkan jenis kelamin
proportion_major <- calculate_proportion(data, "Major")              # Proporsi berdasarkan jurusan
proportion_part_time_job <- calculate_proportion(data, "PartTimeJob") # Proporsi berdasarkan pekerjaan paruh waktu

# Menampilkan hasil proporsi keberhasilan
cat("Proportion of Success (Participation) by Gender:\n")
print(proportion_gender)  # Menampilkan proporsi berdasarkan jenis kelamin

cat("\nProportion of Success (Participation) by Major:\n")
print(proportion_major)    # Menampilkan proporsi berdasarkan jurusan

cat("\nProportion of Success (Participation) by Part-Time Job:\n")
print(proportion_part_time_job)  # Menampilkan proporsi berdasarkan pekerjaan paruh waktu
```

```r
# Menghitung proporsi partisipasi untuk masing-masing kelompok berdasarkan Gender
proportion_gender <- data %>%
  group_by(Gender) %>%
  summarise(
    Proportion_Success = mean(Success),
    N = n()  # Menyimpan jumlah observasi untuk setiap gender
  )

# Menyimpan proporsi dan jumlah untuk Female dan Male
female_proportion <- proportion_gender$Proportion_Success[proportion_gender$Gender == "Female"]
male_proportion <- proportion_gender$Proportion_Success[proportion_gender$Gender == "Male"]
female_n <- proportion_gender$N[proportion_gender$Gender == "Female"]
male_n <- proportion_gender$N[proportion_gender$Gender == "Male"]

# Menghitung perbedaan proporsi
difference_in_proportions <- female_proportion - male_proportion

# Menghitung standar error untuk perbedaan proporsi
std_error <- sqrt((female_proportion * (1 - female_proportion) / female_n) +
                  (male_proportion * (1 - male_proportion) / male_n))

# Z-value untuk 95% CI
z_value <- qnorm(0.975)  # 1.96 untuk 95% CI

# Menghitung Confidence Interval
lower_bound <- difference_in_proportions - z_value * std_error
upper_bound <- difference_in_proportions + z_value * std_error

# Menampilkan hasil
cat("Perbedaan Proporsi antara Female dan Male:", difference_in_proportions, "\n")
cat("95% Confidence Interval for Difference in Proportions: [", lower_bound, ", ", upper_bound, "]\n")
```

```r
# Menghitung proporsi partisipasi untuk masing-masing kelompok berdasarkan Gender
proportion_gender <- data %>%
  group_by(Gender) %>%
  summarise(
    Proportion_Success = mean(Success),
    N = n()  # Menyimpan jumlah observasi untuk setiap gender
  )

# Menyimpan proporsi dan jumlah untuk Female dan Male
female_proportion <- proportion_gender$Proportion_Success[proportion_gender$Gender == "Female"]
male_proportion <- proportion_gender$Proportion_Success[proportion_gender$Gender == "Male"]
female_n <- proportion_gender$N[proportion_gender$Gender == "Female"]
male_n <- proportion_gender$N[proportion_gender$Gender == "Male"]

# Menghitung Relative Risk (RR)
relative_risk <- female_proportion / male_proportion

# Menghitung standar error untuk log(RR)
std_error_log_rr <- sqrt((1 - female_proportion) / (female_n * female_proportion) +
                         (1 - male_proportion) / (male_n * male_proportion))

# Menghitung Confidence Interval untuk log(RR)
z_value <- qnorm(0.975)  # 1.96 untuk 95% CI
log_rr <- log(relative_risk)
ci_lower_log_rr <- log_rr - z_value * std_error_log_rr
ci_upper_log_rr <- log_rr + z_value * std_error_log_rr

# Transformasi kembali ke RR untuk mendapatkan batas bawah dan atas CI
ci_lower_rr <- exp(ci_lower_log_rr)
ci_upper_rr <- exp(ci_upper_log_rr)

# Menampilkan hasil
cat("Relative Risk (Female dibandingkan dengan Male):", relative_risk, "\n")
cat("95% Confidence Interval untuk Relative Risk: [", ci_lower_rr, ", ", ci_upper_rr, "]\n")
```

```r
# Menampilkan hasil Relative Risk dan Confidence Interval
cat("Relative Risk (Female dibandingkan dengan Male):", relative_risk, "\n")
cat("95% Confidence Interval untuk Relative Risk: [", ci_lower_rr, ", ", ci_upper_rr, "]\n")

# Interpretasi hasil
if (relative_risk > 1) {
  cat("Interpretasi: Siswa perempuan (Female) memiliki kemungkinan partisipasi lebih besar dalam kegiatan ekstrakurikuler dibandingkan siswa laki-laki (Male).\n")
} else if (relative_risk < 1) {
  cat("Interpretasi: Siswa perempuan (Female) memiliki kemungkinan partisipasi lebih rendah dalam kegiatan ekstrakurikuler dibandingkan siswa laki-laki (Male).\n")
} else {
  cat("Interpretasi: Tidak ada perbedaan dalam kemungkinan partisipasi antara siswa perempuan (Female) dan siswa laki-laki (Male).\n")
}

# Memeriksa apakah Confidence Interval mencakup nilai 1
if (ci_lower_rr <= 1 && ci_upper_rr >= 1) {
  cat("Kesimpulan: Confidence Interval mencakup nilai 1, yang berarti tidak ada bukti statistik yang cukup untuk menyimpulkan bahwa perbedaan proporsi antara siswa perempuan dan laki-laki signifikan secara statistik pada tingkat kepercayaan 95%.\n")
} else {
  cat("Kesimpulan: Confidence Interval tidak mencakup nilai 1, yang berarti perbedaan proporsi antara siswa perempuan dan laki-laki signifikan secara statistik pada tingkat kepercayaan 95%.\n")
}
```

```r
# Menghitung odds partisipasi untuk setiap gender
odds_gender <- data %>%
  group_by(Gender) %>%
  summarise(
    Success = sum(Success),        # Jumlah partisipasi sukses
    No_Success = n() - sum(Success)  # Jumlah non-partisipasi
  )

# Menghitung odds untuk Female dan Male
odds_female <- odds_gender$Success[odds_gender$Gender == "Female"] / odds_gender$No_Success[odds_gender$Gender == "Female"]
odds_male <- odds_gender$Success[odds_gender$Gender == "Male"] / odds_gender$No_Success[odds_gender$Gender == "Male"]

# Menghitung Odds Ratio (OR)
odds_ratio <- odds_female / odds_male

# Menghitung standar error log(OR)
std_error_log_or <- sqrt(1 / odds_gender$Success[odds_gender$Gender == "Female"] +
                         1 / odds_gender$No_Success[odds_gender$Gender == "Female"] +
                         1 / odds_gender$Success[odds_gender$Gender == "Male"] +
                         1 / odds_gender$No_Success[odds_gender$Gender == "Male"])

# Menghitung Confidence Interval (CI) untuk log(OR)
z_value <- qnorm(0.975)  # 1.96 untuk 95% CI
log_or <- log(odds_ratio)
ci_lower_log_or <- log_or - z_value * std_error_log_or
ci_upper_log_or <- log_or + z_value * std_error_log_or

# Transformasi kembali ke OR untuk mendapatkan batas bawah dan atas CI
ci_lower_or <- exp(ci_lower_log_or)
ci_upper_or <- exp(ci_upper_log_or)

# Menampilkan hasil Odds Ratio (OR) dan Confidence Interval (CI)
cat("Odds Ratio (Female dibandingkan dengan Male):", odds_ratio, "\n")
cat("95% Confidence Interval untuk Odds Ratio: [", ci_lower_or, ", ", ci_upper_or, "]\n")

# Interpretasi hasil
if (odds_ratio > 1) {
  cat("Interpretasi: Siswa perempuan (Female) lebih mungkin berpartisipasi dalam kegiatan ekstrakurikuler dibandingkan siswa laki-laki (Male).\n")
} else if (odds_ratio < 1) {
  cat("Interpretasi: Siswa perempuan (Female) kurang mungkin berpartisipasi dalam kegiatan ekstrakurikuler dibandingkan siswa laki-laki (Male).\n")
} else {
  cat("Interpretasi: Tidak ada perbedaan dalam odds partisipasi antara siswa perempuan (Female) dan siswa laki-laki (Male).\n")
}

# Memeriksa apakah Confidence Interval mencakup nilai 1
if (ci_lower_or <= 1 && ci_upper_or >= 1) {
  cat("Kesimpulan: Confidence Interval mencakup nilai 1, yang berarti tidak ada bukti statistik yang cukup untuk menyimpulkan bahwa perbedaan odds antara siswa perempuan dan laki-laki signifikan secara statistik pada tingkat kepercayaan 95%.\n")
} else {
  cat("Kesimpulan: Confidence Interval tidak mencakup nilai 1, yang berarti perbedaan odds antara siswa perempuan dan laki-laki signifikan secara statistik pada tingkat kepercayaan 95%.\n")
}
```

```r
# Menghitung proporsi partisipasi untuk masing-masing kelompok berdasarkan PartTimeJob
proportion_parttimejob <- data %>%
  group_by(PartTimeJob) %>%
  summarise(Proportion_Success = mean(Success), N = n())  # Menambahkan N untuk setiap kelompok

# Menampilkan proporsi untuk setiap kelompok PartTimeJob
print(proportion_parttimejob)

# Menyimpan proporsi dan jumlah untuk PartTimeJob yang diinginkan
parttime_success <- proportion_parttimejob$Proportion_Success[proportion_parttimejob$PartTimeJob == "Yes"]
not_parttime_success <- proportion_parttimejob$Proportion_Success[proportion_parttimejob$PartTimeJob == "No"]
parttime_n <- proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "Yes"]
not_parttime_n <- proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "No"]

# Menghitung Relative Risk (RR)
relative_risk <- parttime_success / not_parttime_success

# Menghitung standar error untuk log(RR)
std_error_log_rr <- sqrt((1 - parttime_success) / (parttime_n * parttime_success) +
                         (1 - not_parttime_success) / (not_parttime_n * not_parttime_success))

# Menghitung Confidence Interval untuk log(RR)
z_value <- qnorm(0.975)  # 1.96 untuk 95% CI
log_rr <- log(relative_risk)
ci_lower_log_rr <- log_rr - z_value * std_error_log_rr
ci_upper_log_rr <- log_rr + z_value * std_error_log_rr

# Transformasi kembali ke RR untuk mendapatkan batas bawah dan atas CI
ci_lower_rr <- exp(ci_lower_log_rr)
ci_upper_rr <- exp(ci_upper_log_rr)

# Menampilkan hasil
cat("Relative Risk (Pekerjaan Paruh Waktu dibandingkan dengan Tidak):", relative_risk, "\n")
cat("95% Confidence Interval untuk Relative Risk: [", ci_lower_rr, ", ", ci_upper_rr, "]\n")

# Interpretasi Relative Risk (RR)
if (relative_risk > 1) {
  cat("Interpretasi RR: Partisipan dengan pekerjaan paruh waktu lebih mungkin berpartisipasi dalam kegiatan ekstrakurikuler dibandingkan dengan yang tidak bekerja paruh waktu.\n")
} else if (relative_risk < 1) {
  cat("Interpretasi RR: Partisipan dengan pekerjaan paruh waktu kurang mungkin berpartisipasi dalam kegiatan ekstrakurikuler dibandingkan dengan yang tidak bekerja paruh waktu.\n")
} else {
  cat("Interpretasi RR: Tidak ada perbedaan risiko partisipasi antara partisipan dengan pekerjaan paruh waktu dan yang tidak bekerja paruh waktu.\n")
}

# Interpretasi Confidence Interval RR
if (ci_lower_rr <= 1 && ci_upper_rr >= 1) {
  cat("Kesimpulan RR: Confidence Interval mencakup nilai 1, tidak ada bukti statistik yang cukup untuk menyimpulkan perbedaan risiko yang signifikan.\n")
} else {
  cat("Kesimpulan RR: Confidence Interval tidak mencakup nilai 1, perbedaan risiko antara partisipan dengan pekerjaan paruh waktu dan tanpa pekerjaan signifikan secara statistik.\n")
}
```

```r
# Menghitung odds partisipasi untuk setiap kelompok PartTimeJob
odds_parttimejob <- data %>%
  group_by(PartTimeJob) %>%
  summarise(
    Success = sum(Success),         # Jumlah partisipasi sukses
    No_Success = n() - sum(Success)  # Jumlah non-partisipasi
  )

# Menghitung odds untuk Yes dan No
odds_yes <- odds_parttimejob$Success[odds_parttimejob$PartTimeJob == "Yes"] / odds_parttimejob$No_Success[odds_parttimejob$PartTimeJob == "Yes"]
odds_no <- odds_parttimejob$Success[odds_parttimejob$PartTimeJob == "No"] / odds_parttimejob$No_Success[odds_parttimejob$PartTimeJob == "No"]

# Menghitung Odds Ratio (OR)
odds_ratio <- odds_yes / odds_no

# Menghitung standar error log(OR)
std_error_log_or <- sqrt(1 / odds_parttimejob$Success[odds_parttimejob$PartTimeJob == "Yes"] +
                         1 / odds_parttimejob$No_Success[odds_parttimejob$PartTimeJob == "Yes"] +
                         1 / odds_parttimejob$Success[odds_parttimejob$PartTimeJob == "No"] +
                         1 / odds_parttimejob$No_Success[odds_parttimejob$PartTimeJob == "No"])

# Menghitung Confidence Interval (CI) untuk log(OR)
log_or <- log(odds_ratio)
ci_lower_log_or <- log_or - z_value * std_error_log_or
ci_upper_log_or <- log_or + z_value * std_error_log_or

# Transformasi kembali ke OR untuk mendapatkan batas bawah dan atas CI
ci_lower_or <- exp(ci_lower_log_or)
ci_upper_or <- exp(ci_upper_log_or)

# Menampilkan hasil Odds Ratio (OR) dan Confidence Interval (CI)
cat("Odds Ratio (Pekerjaan Paruh Waktu dibandingkan dengan Tidak):", odds_ratio, "\n")
cat("95% Confidence Interval untuk Odds Ratio: [", ci_lower_or, ", ", ci_upper_or, "]\n")

# Interpretasi Odds Ratio (OR)
if (odds_ratio > 1) {
  cat("Interpretasi OR: Partisipan dengan pekerjaan paruh waktu lebih mungkin berpartisipasi dalam kegiatan ekstrakurikuler dibandingkan dengan yang tidak bekerja paruh waktu.\n")
} else if (odds_ratio < 1) {
  cat("Interpretasi OR: Partisipan dengan pekerjaan paruh waktu kurang mungkin berpartisipasi dalam kegiatan ekstrakurikuler dibandingkan dengan yang tidak bekerja paruh waktu.\n")
} else {
  cat("Interpretasi OR: Tidak ada perbedaan dalam odds partisipasi antara partisipan dengan pekerjaan paruh waktu dan yang tidak bekerja paruh waktu.\n")
}

```

```r
# Menghitung proporsi keberhasilan untuk setiap kategori variabel Gender dan PartTimeJob
proportion_gender <- data %>% group_by(Gender) %>%
  summarise(Proportion_Success = mean(Success), N = n())

proportion_parttimejob <- data %>% group_by(PartTimeJob) %>%
  summarise(Proportion_Success = mean(Success), N = n())

# Menampilkan proporsi untuk Gender
print(proportion_gender)
# Menampilkan proporsi untuk PartTimeJob
print(proportion_parttimejob)

# 2) Menghitung perbedaan proporsi untuk Gender dan PartTimeJob
female_success <- proportion_gender$Proportion_Success[proportion_gender$Gender == "Female"]
male_success <- proportion_gender$Proportion_Success[proportion_gender$Gender == "Male"]
parttime_success <- proportion_parttimejob$Proportion_Success[proportion_parttimejob$PartTimeJob == "Yes"]
not_parttime_success <- proportion_parttimejob$Proportion_Success[proportion_parttimejob$PartTimeJob == "No"]

diff_gender <- female_success - male_success
diff_parttime <- parttime_success - not_parttime_success

cat("Difference in proportions (Female - Male):", diff_gender, "\n")
cat("Difference in proportions (Part-time - No Part-time):", diff_parttime, "\n")

# 3) Menghitung 95% CI untuk perbedaan proporsi (Gender dan PartTimeJob)
# Untuk Gender
std_error_gender <- sqrt((female_success * (1 - female_success) / proportion_gender$N[proportion_gender$Gender == "Female"]) +
                         (male_success * (1 - male_success) / proportion_gender$N[proportion_gender$Gender == "Male"]))
ci_lower_gender <- diff_gender - qnorm(0.975) * std_error_gender
ci_upper_gender <- diff_gender + qnorm(0.975) * std_error_gender

cat("95% CI for difference in proportions (Female - Male): [", ci_lower_gender, ", ", ci_upper_gender, "]\n")

# Untuk PartTimeJob
std_error_parttime <- sqrt((parttime_success * (1 - parttime_success) / proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "Yes"]) +
                           (not_parttime_success * (1 - not_parttime_success) / proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "No"]))
ci_lower_parttime <- diff_parttime - qnorm(0.975) * std_error_parttime
ci_upper_parttime <- diff_parttime + qnorm(0.975) * std_error_parttime

cat("95% CI for difference in proportions (Part-time - No Part-time): [", ci_lower_parttime, ", ", ci_upper_parttime, "]\n")

# 4) Menghitung Relative Risk (RR) untuk Gender dan PartTimeJob
rr_gender <- female_success / male_success
rr_parttime <- parttime_success / not_parttime_success

cat("Relative Risk (Female dibandingkan dengan Male):", rr_gender, "\n")
cat("Relative Risk (Part-time dibandingkan dengan No Part-time):", rr_parttime, "\n")

# 5) Menghitung 95% CI untuk RR
# Untuk Gender
std_error_log_rr_gender <- sqrt((1 - female_success) / (proportion_gender$N[proportion_gender$Gender == "Female"] * female_success) +
                                (1 - male_success) / (proportion_gender$N[proportion_gender$Gender == "Male"] * male_success))
log_rr_gender <- log(rr_gender)
ci_lower_log_rr_gender <- log_rr_gender - qnorm(0.975) * std_error_log_rr_gender
ci_upper_log_rr_gender <- log_rr_gender + qnorm(0.975) * std_error_log_rr_gender
ci_lower_rr_gender <- exp(ci_lower_log_rr_gender)
ci_upper_rr_gender <- exp(ci_upper_log_rr_gender)

cat("95% CI for RR (Female vs Male): [", ci_lower_rr_gender, ", ", ci_upper_rr_gender, "]\n")

# Untuk PartTimeJob
std_error_log_rr_parttime <- sqrt((1 - parttime_success) / (proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "Yes"] * parttime_success) +
                                  (1 - not_parttime_success) / (proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "No"] * not_parttime_success))
log_rr_parttime <- log(rr_parttime)
ci_lower_log_rr_parttime <- log_rr_parttime - qnorm(0.975) * std_error_log_rr_parttime
ci_upper_log_rr_parttime <- log_rr_parttime + qnorm(0.975) * std_error_log_rr_parttime
ci_lower_rr_parttime <- exp(ci_lower_log_rr_parttime)
ci_upper_rr_parttime <- exp(ci_upper_log_rr_parttime)

cat("95% CI for RR (Part-time vs No Part-time): [", ci_lower_rr_parttime, ", ", ci_upper_rr_parttime, "]\n")

# 6) Interpretasi untuk RR
cat("Interpretasi RR Gender: Jika RR > 1, risiko partisipasi untuk perempuan lebih tinggi dibandingkan dengan laki-laki.\n")
cat("Interpretasi RR PartTimeJob: Jika RR > 1, risiko partisipasi untuk yang bekerja paruh waktu lebih tinggi dibandingkan dengan yang tidak.\n")

# 7) Menghitung Odds Ratio (OR)
# Untuk Gender
odds_female <- female_success / (1 - female_success)
odds_male <- male_success / (1 - male_success)
or_gender <- odds_female / odds_male

# Untuk PartTimeJob
odds_parttime <- parttime_success / (1 - parttime_success)
odds_not_parttime <- not_parttime_success / (1 - not_parttime_success)
or_parttime <- odds_parttime / odds_not_parttime

cat("Odds Ratio (Female dibandingkan dengan Male):", or_gender, "\n")
cat("Odds Ratio (Part-time dibandingkan dengan No Part-time):", or_parttime, "\n")

# 8) Menghitung 95% CI untuk OR
# Untuk Gender
std_error_log_or_gender <- sqrt(1 / (proportion_gender$N[proportion_gender$Gender == "Female"] * female_success) +
                                1 / (proportion_gender$N[proportion_gender$Gender == "Female"] * (1 - female_success)) +
                                1 / (proportion_gender$N[proportion_gender$Gender == "Male"] * male_success) +
                                1 / (proportion_gender$N[proportion_gender$Gender == "Male"] * (1 - male_success)))
log_or_gender <- log(or_gender)
ci_lower_log_or_gender <- log_or_gender - qnorm(0.975) * std_error_log_or_gender
ci_upper_log_or_gender <- log_or_gender + qnorm(0.975) * std_error_log_or_gender
ci_lower_or_gender <- exp(ci_lower_log_or_gender)
ci_upper_or_gender <- exp(ci_upper_log_or_gender)

cat("95% CI for OR (Female vs Male): [", ci_lower_or_gender, ", ", ci_upper_or_gender, "]\n")

# Untuk PartTimeJob
std_error_log_or_parttime <- sqrt(1 / (proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "Yes"] * parttime_success) +
                                  1 / (proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "Yes"] * (1 - parttime_success)) +
                                  1 / (proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "No"] * not_parttime_success) +
                                  1 / (proportion_parttimejob$N[proportion_parttimejob$PartTimeJob == "No"] * (1 - not_parttime_success)))
log_or_parttime <- log(or_parttime)
ci_lower_log_or_parttime <- log_or_parttime - qnorm(0.975) * std_error_log_or_parttime
ci_upper_log_or_parttime <- log_or_parttime + qnorm(0.975) * std_error_log_or_parttime
ci_lower_or_parttime <- exp(ci_lower_log_or_parttime)
ci_upper_or_parttime <- exp(ci_upper_log_or_parttime)

cat("95% CI for OR (Part-time vs No Part-time): [", ci_lower_or_parttime, ", ", ci_upper_or_parttime, "]\n")

# 9) Interpretasi untuk OR
cat("Interpretasi OR Gender: Jika OR > 1, odds partisipasi untuk perempuan lebih tinggi dibandingkan dengan laki-laki.\n")
cat("Interpretasi OR PartTimeJob: Jika OR > 1, odds partisipasi untuk yang bekerja paruh waktu lebih tinggi dibandingkan dengan yang tidak.\n")
```
